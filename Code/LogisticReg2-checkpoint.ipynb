# Create a logistic regression model to predict the default rates

**This notebook aims to build the full model aswell as test it against the train and test data. We will also test all asumptions aswell as validate the model in order too see if it's predictions is actually significant**

## Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import statsmodels.formula.api as smf 
import statsmodels.api as sm 
import statsmodels.genmod.families.links as lnk 
import statsmodels.genmod.families as fam

from scipy.stats import chi2 
from scipy.stats import rankdata


import patsy
from statsmodels.stats.outliers_influence import variance_inflation_factor

from statsmodels.nonparametric.smoothers_lowess import lowess

from scipy.stats import rankdata

from sklearn.metrics import roc_curve, auc, confusion_matrix, ConfusionMatrixDisplay  
```

## We need too look at what we want to predict

* We want to predict if someone will default.

* This is a binary value, seeing as a 1 = default and 0 = non-default

* Thus we will use the model to predict the binary default variable from all  our other variables

## Import and inspect data

```python
train_df = pd.read_csv("Credit_Score_Train.csv")

# Drop unnecessary columns
train_df.drop("Unnamed: 0", axis=1, inplace=True)
train_df.drop("X", axis=1, inplace=True)

# Inspect dataframe
train_df.head()
```

```python
train_df.describe()
```

**We can clearly see in most columns that there is no outliers, however in the Income and Loan Amount columns we see that we have some very big numbers. We cannot just use the normal outlier detection method as we know that these two fields are usually skewed to the right. Thus because of this we are firs going to preform a log transform and form that data check if we have any outliers**

```python
# Log-transform both income and LoanAmt in order to remove outliers
train_df["log_Income"] = np.log1p(train_df["Income"])
train_df["log_LoanAmt"] = np.log1p(train_df["LoanAmt"])

# Check for outliers in these columns 
bounds = {}

# log_Income
q1, q3 = train_df["log_Income"].quantile([0.25, 0.75])
iqr = q3 - q1
bounds["log_Income"] = (q1 - 4 * iqr, q3 + 2 * iqr)

# log_LoanAmt
q1, q3 = train_df["log_LoanAmt"].quantile([0.25, 0.75])
iqr = q3 - q1
bounds["log_LoanAmt"] = (q1 - 4 * iqr, q3 + 2 * iqr)

# Get number of outliers that was present and remove them

# Count outliers
for col, (low, high) in bounds.items():
    n_low = (train_df[col] < low).sum()
    n_high = (train_df[col] > high).sum()
    print(f"{col}: {n_low} low, {n_high} high, bounds = ({low:.2f}, {high:.2f})")


mask = pd.Series(True, index=train_df.index)
for col, (low, high) in bounds.items():
    mask &= train_df[col].between(low, high)

train_clean = train_df[mask].copy()
n_after = train_clean.shape[0]

print(n_after)
print(f"Removed {len(train_df) - len(train_clean)} rows")
train_clean.describe()
```

**Possible multicolinearity**

```python
fig, axes = plt.subplots(2,1, figsize=(7,12))

axes[0].scatter(train_clean["Income"], train_clean["LoanAmt"])
axes[0].set_xlabel("Income")
axes[0].set_ylabel("Loan Amount")
axes[0].set_title("Loan amount vs income", fontsize=12, fontweight='bold')

axes[1].scatter(train_clean["HistLen"], train_clean["Age"])
axes[1].set_title("History length vs Age", fontsize=12, fontweight="bold")
axes[1].set_xlabel("History length")
axes[1].set_ylabel("Age")
fig.suptitle("Possible multicolinearity", fontsize=20, fontweight='bold')
plt.show()
```

## Building Model 1

```python
logist = smf.glm( 
formula="default ~IntRate + log_LoanAmt + log_Income + HistLen +  C(Offences) + Utilisation + DTI + Age + " 
"C(Purpose) + C(EmpStatus) + C(HomeOwn) + C(Gender) + C(Education)", 
family=fam.Binomial(link=lnk.Logit()), 
data=train_clean, 
).fit() 

logist.summary()
```

# Check validity of model

**a) Now we can check if the model is valid or not with residual deviation statistic**

```python
# H0: Model fits vs Ha: Model does not fit

dfd = logist.df_resid 
devstat = logist.deviance 
pval = 1 - chi2.cdf(devstat, df=dfd) 

print(f"\nDeviance={devstat}\ndf={dfd}\np-value={pval}\n") 

if pval < 0.05:
    print("Reject H0. Model does not fit")
else:
    print("Do not reject H0. Model does fit")
```

This test might not correct seeing as defaults are binary and there are several continuous variables that might skew the result of this test

**b) Compare current model with no model. Use the likelihood ratio test**

* H0: all 21 slope coefficients are zero
* Ha: at least one is non-zero

```python
lr = 2*(logist.llf - logist.llnull) # Test statistic
print(f"likelyhood-ratio: {lr:.4f} ,\nDegrees of freedom: {logist.df_model} ,\np-value: {chi2.sf(lr, logist.df_model):.4f}")
```

We reject H0, meaning this model is better than no model at all

**c) Do the Hosmer-Lemeshow goodness-of-fit test**

This will act as a better goodness-of-fit test than the residual deviance statistic, as it will group the data by binning observations based on predicted probability

```python
'''
H0: The model is well calibrated
Ha: The model is not well calibrated
'''

y = logist.model.endog # predicted outcomes (0/1)
p = logist.predict() # Predicted probabilities
g = 10 # Groups

d = pd.DataFrame({"y": y, "p": p})
d["bin"] = pd.qcut(d['p'], g, duplicates='drop') # cut into quantiles

t = d.groupby('bin', observed=True).agg(obs=("y", "sum"), exp=("p","sum"), n=("y","count"))
HL = (((t.obs - t.exp)**2) / (t.exp * (1 - t.exp/t.n))).sum() # follow chi-square
df = len(t) - 2 
p_value = 1 - chi2.cdf(HL, df)

print(f"Hosmer-Lemeshow: \n* Statistic: {HL:.4f} \n* Degrees of freedom: {df} \n* p-value: {p_value:.4f}")
```

We cannot reject H0, which means there is no evedance that our model has discrimination in

**c) Make a calibration plot to visualise this**

```python
fig, axe = plt.subplots(1,1,figsize=(5,5))

fig.suptitle("Calibration plot too show goodness of fit", fontsize=16, fontweight="bold")
axe.plot(t.exp/t.n, t.obs/t.n, '-o', color="#C56A2D", linewidth=2.5, markersize=7, markeredgewidth=1.5)
axe.plot([0,1],[0,1], '--', color="#6A4C93", linewidth=1.8, label="Perfect calibration")

axe.set_xlabel("Expected proportion", fontsize=12)
axe.set_ylabel("Observed proportion", fontsize=12)

axe.legend(fontsize=10)

fig.tight_layout()
fig.savefig("calibration_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

**The graph shows a good fit to the prefect calibration. This shows that we have good calibration**

**d) Get the AUC and Brier score**

```python
# Brier score: mean squared error of the probabilities
brier = np.mean((p - y) ** 2)

# AUC via the rank-sum (Mann-Whitney) formula
ranks = rankdata(p)
n1 = y.sum()
n0 = len(y) - n1
auc = (ranks[y == 1].sum() - n1 * (n1 + 1) / 2) / (n1 * n0)

print(f"AUC  : {auc:.4f}")
print(f"Brier: {brier:.4f}")
```

* Some discrimination si shown by the AUC
* On the line between moderate and good calibration

# Check for linearity

**a) Spline likelihood ratio test**

```python
cats = "C(Purpose) + C(EmpStatus) + C(HomeOwn) + C(Gender) + C(Education)"
cont = ["IntRate", "log_Income", "LoanToIncome", "Utilisation", "DTI", "Offences"]

base = smf.glm("default ~ " + " + ".join(cont) + " + " + cats, family=fam.Binomial(), data=train_clean).fit()

for v in cont:
    new = f"C({v})" if v == "Offences" else f"bs({v}, df=4)"
    terms = [new if t == v else t for t in cont]
    m = smf.glm("default ~ " + " + ".join(terms) + " + " + cats,
        family=fam.Binomial(), data=train_clean).fit()
    lr = 2 * (m.llf - base.llf)
    d = m.df_model - base.df_model
    print(f"{v:14s} LR={lr:6.2f}  df={int(d)}  p={chi2.sf(lr, d):.4f}  dAIC={m.aic - base.aic:6.1f}")
```

We can see Utilisation does not follow a linear relationship. Thus splines will work better for it. We will see from the deleted residual plots which relationship it has to the data

**b) Partial residual plot of Utilisation**

```python
x = train_clean["Utilisation"].values
pr = base.resid_working + base.params["Utilisation"] * x # get partial residuals

"""
https://www.statsmodels.org/devel/generated/statsmodels.nonparametric.smoothers_lowess.lowess.html

LOWESS (Locally Weighted Scatterplot Smoothing)

A lowess function that outputs smoothed estimates of endog at the given exog values from points (exog, endog)
"""
sm_line = lowess(pr, x, frac=0.4)

# binned means: easier to read than the raw defaults
d = pd.DataFrame({"x": x, "pr": pr})
d["bin"] = pd.qcut(d["x"], 20, duplicates="drop")
b = d.groupby("bin", observed=True).agg(x=("x", "mean"), pr=("pr", "mean"))

# straight line the model assumes (slope = fitted coefficient), aligned to the data
x_sort = np.sort(x)
offset = pr.mean() - base.params["Utilisation"] * x.mean()
lin = base.params["Utilisation"] * x_sort + offset
```

Plot the partial residuals.

```python
fig, ax = plt.subplots(figsize=(8, 5))
fig.suptitle("Partial residual plot for utilisation", fontsize=15, fontweight='bold')
ax.scatter(x, pr, s=2, alpha=0.1, color="grey")
ax.plot(x_sort, lin, "k--", label="Linear (model assumption)")
ax.plot(sm_line[:, 0], sm_line[:, 1], "r-", lw=2, label="Lowess (data)")
ax.plot(b.x, b.pr, "bo", label="Bin means (20 bins)")
ax.set_xlabel("Utilisation"); ax.set_ylabel("Partial residual")
ax.set_ylim(np.percentile(pr, 2), np.percentile(pr, 98))   # trim the extreme tails of the cloud
ax.legend(); 
plt.show()
```

The LOWESS curve shows that Utilisation is close to the straight line. Seeing as the p-value still meet the 1% level to not reject H0 in the linearity test, we can say because of evedince from the graph, keeping utilisation linear will be efficient.

# Simplyfy model with the joint

**a) Do wald test to check for insignificance**

```python
full = smf.glm("default ~ IntRate + log_Income + LoanToIncome + C(Offences) + Utilisation + DTI + "
               "C(Purpose) + C(EmpStatus) + C(HomeOwn) + C(Gender) + C(Education)",
               family=fam.Binomial(), data=train_clean).fit()

print(full.wald_test_terms(scalar=True))
print(full.wald_test("log_Income = 0, LoanToIncome = 0", scalar=True))
```

**b) Rebuild model**

```python
# Stage 1: the clearly useless variables
m1 = smf.glm("default ~ IntRate + Utilisation + DTI + Offences + C(EmpStatus) + C(Gender) + C(Education)",
             family=fam.Binomial(), data=train_clean).fit()
print(m1.wald_test_terms(scalar=True))

# Stage 2: the weak ones, if they are still insignificant in m1
final = smf.glm("default ~ IntRate + Utilisation + DTI + C(EmpStatus)",
                family=fam.Binomial(), data=train_clean).fit()
```

*Check how the new model fits*

```python
# Overall check: full vs final
lr = 2 * (full.llf - final.llf)
d = int(full.df_model - final.df_model)
print(f"Full vs final: LR={lr:.2f}, df={d}, p={chi2.sf(lr, d):.4f}")
print("AIC:", full.aic, final.aic, "  BIC:", full.bic, final.bic)
print(final.summary())
```

*Check how the almost correct model fits*

```python
# Overall check: final vs m1
lr = 2 * (m1.llf - final.llf)
d = int(m1.df_model - final.df_model)
print(f"Full vs final: LR={lr:.2f}, df={d}, p={chi2.sf(lr, d):.4f}")
print("AIC:", final.aic, m1.aic, "  BIC:", final.bic, m1.bic)
print(final.summary())
```

There is no significant difference between the final model and the model with almost all significant values, thus we use final as our model

# Run model on the testing set

```python
test_df = pd.read_csv("test_scorecard_dataset.csv")
test_df.drop("Unnamed: 0", axis=1, inplace=True)
test_df.head()
```

**a) Predict the outcomes**

```python
p_test = final.predict(test_df)
y_test = test_df["default"]. values
assert len(p_test) == len(test_df) 

p_train = final.predict(train_clean)
y_train = train_clean["default"].values
```

```python
# Brier score: mean squared error of the probabilities
brier_train = np.mean((p_train - y_train) ** 2)

# AUC via the rank-sum (Mann-Whitney) formula
ranks = rankdata(p_train)
n1 = y_train.sum()
n0 = len(y_train) - n1
auc_train = (ranks[y_train == 1].sum() - n1 * (n1 + 1) / 2) / (n1 * n0)

# Brier score: mean squared error of the probabilities
brier_test = np.mean((p_test - y_test) ** 2)

# AUC via the rank-sum (Mann-Whitney) formula
ranks = rankdata(p_test)
n1 = y_test.sum()
n0 = len(y_test) - n1
auc_test = (ranks[y_test == 1].sum() - n1 * (n1 + 1) / 2) / (n1 * n0)

```

```python
print(f"Train: AUC={auc_train:.3f}  Brier={brier_train:.4f}")
print(f"Test: AUC={auc_test:.3f}  Brier={brier_test:.4f}")
```

**Draw the ROC curve**

```python
fpr_train, tpr_train, threshold_train = roc_curve(y_train, p_train) 
roc_auc = auc(fpr_train, tpr_train) 

fpr_test, tpr_test, threshold_test = roc_curve(y_test, p_test) 
roc_auc = auc(fpr_test, tpr_test) 

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

fig.suptitle("ROC curves for both test and train data", fontsize=17, fontweight='bold')

axes[0].plot(fpr_train, tpr_train, label=f"ROC curve (AUC = {roc_auc:.3f})")
axes[0].set_title("Train data", fontsize=12, fontweight='bold')
axes[0].plot([0, 1], [0, 1], linestyle="--") 
axes[0].set_xlabel("False Positive Rate") 
axes[0].set_ylabel("True Positive Rate") 
axes[0].legend(loc="lower right") 



axes[1].plot(fpr_test, tpr_test, label=f"ROC curve (AUC = {roc_auc:.3f})") 
axes[1].set_title("Test data", fontsize=12, fontweight='bold')
axes[1].plot([0, 1], [0, 1], linestyle="--") 
axes[1].set_xlabel("False Positive Rate") 
axes[1].set_ylabel("True Positive Rate") 
axes[1].legend(loc="lower right") 

plt.savefig("ROC_curve.png", dpi=300, bbox_inches="tight")
plt.show() 
```

```python
fpr, tpr, thr = roc_curve(y_train, p_train)
cutoff = thr[np.argmax(tpr - fpr)]
print("Cutoff:", round(cutoff, 3))

pred = (p_test >= cutoff).astype(int)
ConfusionMatrixDisplay(confusion_matrix(y_test, pred), display_labels=["No default", "Default"]).plot(cmap="Blues")
plt.savefig("confusion_matrix.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
y_tr, p_tr = train_clean["default"].values, final.predict(train_clean).values
y_te, p_te = test_df["default"].values,     final.predict(test_df).values

# Cutoff chosen on training data only, then applied unchanged to the test set
fpr, tpr, thr = roc_curve(y_tr, p_tr)
cutoff = thr[np.argmax(tpr - fpr)]
print(f"Cutoff (Youden, training) = {cutoff:.3f}\n")

def five_metrics(y, p, c):
    tn, fp, fn, tp = confusion_matrix(y, (p >= c).astype(int)).ravel()
    return {"Accuracy":    (tp + tn) / (tp + tn + fp + fn),
            "Sensitivity": tp / (tp + fn),
            "Specificity": tn / (tn + fp),
            "Precision":   tp / (tp + fp),
            "F1":          2 * tp / (2 * tp + fp + fn)}

res = pd.DataFrame({"Train": five_metrics(y_tr, p_tr, cutoff),
                    "Test":  five_metrics(y_te, p_te, cutoff)})
print(res.round(3))

res.plot.bar(figsize=(7, 4), rot=0)
plt.ylim(0, 1); plt.ylabel("Value")
plt.title(f"Classification metrics at cutoff {cutoff:.3f}")
plt.tight_layout()
plt.savefig("metrics_bar.png", dpi=300, bbox_inches="tight")
plt.show()
```
