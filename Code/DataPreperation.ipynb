## Identify messy data

**1. ID should be a string**

```python
biographical["ID"] = biographical["ID"].astype(str)
biographical.dtypes
```

```python
biographical.head()
```

```python
biographical["Education"].value_counts()
```

**2. Check if all offences is of numeric type**

```python
biographical['Offences'] = pd.to_numeric(biographical['Offences'], errors='coerce')
null_offence = pd.isnull(biographical['Offences']).sum()

if null_offence == 0:
    print("There are 0 Na occurances which means all values are numeric")
else:
    print("There are Na values. Inspect further.")
```

**3. Check for missing values**

```python
biographical.isnull().sum()
```

**4.1. Check for spelling mistakes in Gender**

```python
valid_Genders = ["Male", "Female"]
invalid_Gender_idx = biographical[~biographical["Gender"].isin(valid_Genders)] 
if (len(invalid_Gender_idx) == 0):
    print("No spelling errors in Gender")
```

**4.2. Check for spelling mistakes in Education**

```python
valid_Education = ["Grade 12", "Degree", "Diploma", "Postgrad", "No records"]
invalid_Education_idx = biographical[~biographical["Education"].isin(valid_Education)] 
if (len(invalid_Education_idx) == 0):
    print("No spelling errors in Education")
else:
    print("Inspect: "+ str(len(invalid_Education_idx)) + " errors" )
```
