## Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Get data into a pandas dataframe

```python
biographical = pd.read_excel("01 scorecard_dataset.xlsx")
biographical.head()
```

```python
credit_df = pd.read_excel("01 scorecard_dataset.xlsx", sheet_name=1)
for i in range(2,10):
    credit_df = pd.concat([credit_df, pd.read_excel("01 scorecard_dataset.xlsx", sheet_name=i)], ignore_index=True)

credit_df.head()
```

## Save as .csv files

```python
biographical.to_csv("Biographical_raw.csv", index=False)
credit_df.to_csv("Credit_data_raw.csv", index=False)
```
